```python
import numpy as np
import pandas as pd
```

activation function

```python
def Relu(x):
  #return x if x>0 else np.array([0])
  return np.maximum(0, x)
def lin_grad(yx,yz):
  return 2*(yx-yz)
```

```python
input=np.array([1])
hidden=np.array([0.4,-0.2])
output=np.array([[0.6],[-0.1]])
```

# forward pass

```python
def hidden_compute(inp,hid):
  compute=[]
  for val in hid:
    compute.append(Relu(val*inp))
  return np.array(compute)


```

```python
print(hidden_compute(input,hidden))
```

```python
def output_compute(hid,out):
  final=hid*out
  return np.sum(final)
```

```python
hidd=hidden_compute(input,hidden)
print(output_compute(hidd,output))
```
