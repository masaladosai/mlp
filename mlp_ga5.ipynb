```python
import pandas as pd
from google.colab import files
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,Ridge,Lasso
```

```python
uploaded_file=files.upload()
filename=list(uploaded_file)[0]
df=pd.read_csv(filename)
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
train_x,test_x,train_y,test_y=train_test_split(df,df["Credit_Limit"],test_size=0.3,random_state=42)
```

```python
lr=LinearRegression(fit_intercept=False)
lr.fit(train_x,train_y)
```

```python
from sklearn.metrics import classification_report,r2_score
lr_pred=lr.predict(test_x)
print(r2_score(test_y,lr_pred))
```

```python
import numpy as np

coef = lr.coef_
idx = np.argmax(np.abs(coef))

print(idx)
```

```python
r_model = Ridge(solver='sag', tol=0.0005, random_state=42)
r_model.fit(train_x,train_y)
```

```python
r_pred=r_model.predict(test_x)
print(r2_score(test_y,r_pred))
```

```python
print(f"{r_model.intercept_:.3f}")
```

```python
lasso_model = Lasso(alpha=100, random_state=42)
lasso_model.fit(train_x, train_y)

lasso_pred = lasso_model.predict(test_x)
r2 = r2_score(test_y, lasso_pred)

print(round(r2, 3))
```

```python
coef = lasso_model.coef_
count = np.sum((coef >= -1) & (coef <= 1))

print(count)
```

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error

model = KNeighborsRegressor(n_neighbors=10, p=1)
model.fit(train_x, train_y)

y_pred = model.predict(test_x)

rmse = np.sqrt(mean_squared_error(test_y, y_pred))
print(round(rmse, 3))
```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(
    max_depth=10,
    min_samples_split=6,
    min_samples_leaf=6,
    random_state=42
)

model.fit(train_x, train_y)

y_pred = model.predict(test_x)

rmse = np.sqrt(mean_squared_error(test_y, y_pred))
print(round(rmse, 3))
```

```python
from sklearn.ensemble import AdaBoostRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import r2_score

# model
model = AdaBoostRegressor(random_state=42)

# param grid
param_grid = {
    'n_estimators': [10, 50, 100, 200, 500],
    'learning_rate': [0.1, 0.5, 1, 2]
}

# grid search
grid = GridSearchCV(model, param_grid, cv=4)
grid.fit(train_x, train_y)

# best model
best_model = grid.best_estimator_

# predictions
y_pred = best_model.predict(test_x)

# r2 score
r2 = r2_score(test_y, y_pred)
print(round(r2, 3))
```
